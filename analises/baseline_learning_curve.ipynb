```python
import os
import numpy as np

def find_learning_curves(root_folder):
    """
    Busca recursivamente por arquivos 'learning_process.txt' em uma pasta.
    (Esta função pode ser reutilizada do script anterior se estiver na mesma sessão)
    """
    filepaths = []
    for dirpath, _, filenames in os.walk(root_folder):
        if 'learning_process.txt' in filenames:
            filepaths.append(os.path.join(dirpath, 'learning_process.txt'))
    return filepaths

def calculate_global_average_performance(filepaths):
    """
    Carrega todas as curvas de aprendizado e calcula a média geral de todos os pontos.

    Args:
        filepaths (list): Lista de caminhos para os arquivos de curva de aprendizado.

    Returns:
        tuple: Uma tupla contendo (média_geral, desvio_padrão_geral, total_de_pontos).
               Retorna (None, None, 0) se nenhum dado for encontrado.
    """
    if not filepaths:
        return None, None, 0

    # Carrega os dados de cada arquivo e os concatena em um único array 1D
    all_data_points = np.concatenate([np.loadtxt(fp, delimiter=',') for fp in filepaths])
    
    if all_data_points.size == 0:
        return None, None, 0

    # Calcula a média e o desvio padrão de todos os pontos de dados juntos
    global_mean = np.mean(all_data_points)
    global_std = np.std(all_data_points)
    total_points = all_data_points.size
    
    return global_mean, global_std, total_points
```

```python

# --- CÉLULA DE EXECUÇÃO ---

# 1. ESPECIFIQUE O CAMINHO DA PASTA AQUI
# Use o mesmo caminho que você usou para a plotagem dos gráficos.
folder_to_analyze = "../data/models/baseline_repulsivo"

# 2. EXECUTE A ANÁLISE GERAL
if not os.path.isdir(folder_to_analyze):
    print(f"ERRO: A pasta especificada não foi encontrada: '{folder_to_analyze}'")
else:
    # Encontra todos os arquivos de curva de aprendizado
    filepaths = find_learning_curves(folder_to_analyze)
    
    if not filepaths:
        print(f"Nenhum arquivo 'learning_process.txt' encontrado em '{folder_to_analyze}'.")
    else:
        print(f"Analisando {len(filepaths)} arquivos de curva de aprendizado...")
        
        # Calcula as métricas gerais
        mean_perf, std_perf, num_points = calculate_global_average_performance(filepaths)
        
        if mean_perf is not None:
            print("\n--- Desempenho Médio Geral ---")
            print(f"Analisados {num_points} episódios no total.")
            print(f"Média Geral dos Passos Normalizados: {mean_perf:.4f}")
            print(f"Desvio Padrão Geral: {std_perf:.4f}")
            print("\nSignificado: Em média, em todas as simulações e todos os episódios,")
            print(f"o agente levou {mean_perf:.2%} do tempo máximo de um 'trial' para encontrar o alvo.")
        else:
            print("Não foi possível calcular a média. Nenhum dado encontrado nos arquivos.")
```
