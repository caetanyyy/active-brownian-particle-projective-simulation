```python
import os
import numpy as np
import pandas as pd
import altair as alt
import itertools
import sys
sys.path.insert(0, '..')
from agents import PsAgent
```

# Carregando dados

```python
def load_learning_curve(path_modelo):
    learning_curve = np.loadtxt(f'{path_modelo}/learning_process.txt')
    learning_curve = pd.DataFrame(
        learning_curve, columns = ['tempo']
    ).reset_index().rename(columns = {'index':'episodio'})
    return learning_curve

def load_h_matrix(grupos, grupo, modelo, agent):
    if grupos[grupo]['tipo'] == 'colisao':
        matriz = []
        for state, colision, timer in itertools.product(*[[0,1],[0,1],[i for i in range(2000)]]):
            troca, nao_troca = return_percept(agent, [state, timer, colision])
            _ = {
                    'modelo':modelo,
                    'periodo':timer,
                    'state':state,
                    'colision':colision,
                    'troca':troca,
                    'nao_troca':nao_troca,
                }
            matriz.append(_)
        return matriz
    else:
        matriz = []
        for state, timer in itertools.product(*[[0,1],[i for i in range(2000)]]):
            troca, nao_troca = return_percept(agent, [state, timer])
            _ = {
                    'modelo':modelo,
                    'periodo':timer,
                    'colision':np.nan,
                    'state':state,
                    'troca':troca,
                    'nao_troca':nao_troca,
                }
            matriz.append(_)
        return matriz

def return_percept(agent, observation):
    percept = agent.percept_preprocess(observation)

    h_vector = agent.h_matrix[:, percept]
    sum_ = np.sum(h_vector)
    probs = h_vector/sum_

    nao_troca = probs[0]
    troca = probs[1]

    return troca, nao_troca

path = '../data/models'
lista_modelos = os.listdir(path)

grupos = {
    'colision_R=0.001_ref=1':{
        'tipo':'colisao',
        'reward':0.01,
        'reflection':1
        },
    'colision_R=0.01':{
        'tipo':'colisao',
        'reward':0.01,
        'reflection':0
        },
    'colision_R=0.005_ref=1':{
        'tipo':'colisao',
        'reward':0.005,
        'reflection':1
        },
    'colision_R=0.005':{
        'tipo':'colisao',
        'reward':0.005,
        'reflection':0
        },
    'colision_R=1':{
        'tipo':'colisao',
        'reward':1,
        'reflection':0
        },
    'periodico':{
        'tipo':'periodico',
        'reward':np.nan,
        'reflection':0
        },
    'periodico_ref=1':{
        'tipo':'periodico',
        'reward':np.nan,
        'reflection':1
        }
}

modelos = []
args = []
learning_curves = pd.DataFrame()
h_matrix = []

for grupo in list(grupos.keys()):
    lista = os.listdir(f'{path}/{grupo}')
    for modelo in lista:
        path_modelo = f'{path}/{grupo}/{modelo}'
        modelos.append({'modelo':modelo, 'grupo':grupo, 'path':path_modelo})
        # Agente
        agent = PsAgent.load(f'{path_modelo}')

        ##########################################
        # Carregando dados de curva de aprendizado
        learning_curve = load_learning_curve(path_modelo)
        learning_curve['modelo'] = modelo
        learning_curves = pd.concat([learning_curve, learning_curves], axis = 0)

        ##########################################
        # Carregando matriz 
        h_matrix.extend(load_h_matrix(grupos, grupo, modelo, agent))
```

```python
modelos = pd.DataFrame(modelos).merge(pd.DataFrame(grupos).T.reset_index().rename(columns = {'index':'grupo'}))
h_matrix = pd.DataFrame(h_matrix).merge(modelos[['modelo','tipo','reward','reflection']])
learning_curves = pd.DataFrame(learning_curves).merge(modelos[['modelo','tipo','reward','reflection']])
```

```python
from scipy.stats import iqr

learning_curves['reward'] = learning_curves['reward'].fillna(1)
learning_curves['episodio_grupo'] = (learning_curves['episodio']//10)*10
learning_curves_stats = learning_curves.groupby(['tipo','reward','reflection','episodio_grupo'])['tempo'].agg(['mean','std',iqr]).reset_index()
```

```python
db = h_matrix.copy(deep = True)
db['colision'] = db['colision'].fillna(0)
db['reward'] = db['reward'].fillna(1)

db = db.sort_values(['tipo','modelo','state','colision','periodo']).reset_index(drop = True)

db['prod_nao_troca'] = db.groupby(['modelo','state','colision'])['nao_troca'].cumprod().shift(1)
db['P_T_troca'] = db['prod_nao_troca']*db['troca']
db['sum_P_T_troca'] = db.groupby(['modelo','state','colision'])['P_T_troca'].cumsum().shift(1)

db_mean = db.groupby(['tipo','state','colision','reflection','reward','periodo'])[
    ['troca', 'P_T_troca', 'sum_P_T_troca']
].mean().reset_index()

db_mean.head()

db_mean = db_mean.query('periodo > 2 and periodo < 500').reset_index(drop = True)

db_mean['periodo_tao'] = db_mean['periodo']*(1e-4)
```

```python
db_plot = db_mean.query('tipo == "periodico" and reflection == 0')
alt.Chart(db_plot).mark_line(opacity = 0.7).encode(
    x = 'periodo_tao',
    y = alt.Y('sum_P_T_troca').axis(format = '%'),
    row = 'state:N'
).properties(height = 100).resolve_scale(y = 'independent')
```
