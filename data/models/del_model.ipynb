```python
import os

# --- IMPORTANTE ---
# Substitua "coloque_o_caminho_da_sua_pasta_aqui" pelo caminho da pasta que você deseja limpar.
# Exemplo no Windows: "C:\\Users\\SeuUsuario\\Documentos\\PastaParaLimpar"
# Exemplo no macOS/Linux: "/home/SeuUsuario/Documentos/PastaParaLimpar"
caminho_da_pasta = 'maze_P=100'
print(f"Verificando o conteúdo da pasta: {caminho_da_pasta}\n")

if os.path.exists(caminho_da_pasta):
    # os.walk() percorre a árvore de diretórios de cima para baixo.
    # Para cada diretório (incluindo o inicial), ele retorna:
    # 1. O caminho do diretório (root)
    # 2. Uma lista de subdiretórios nesse caminho (dirs)
    # 3. Uma lista de arquivos nesse caminho (files)
    for root, dirs, files in os.walk(caminho_da_pasta):
        print(f"Analisando o diretório: {root}")
        
        if not dirs and not files:
            print("  -> Este diretório está vazio.")
        
        # Lista as subpastas
        if dirs:
            print(f"  Subpastas encontradas: {dirs}")
            
        # Lista os arquivos
        if files:
            print(f"  Arquivos encontrados: {files}")
        
        print("-" * 20)
else:
    print("O caminho especificado não existe. Por favor, verifique.")
```

```python
def del_data(caminho_da_pasta):
    import os
    import shutil

    # --- IMPORTANTE ---
    # Verifique se este é o mesmo caminho que você usou na célula anterior.

    if os.path.exists(caminho_da_pasta):
        print(f"Iniciando a limpeza da pasta: {caminho_da_pasta}\n")

        # os.listdir() retorna uma lista com todos os itens (arquivos e pastas) no diretório.
        for item in os.listdir(caminho_da_pasta):
            item_path = os.path.join(caminho_da_pasta, item)

            try:
                # Se for um arquivo ou um link simbólico, apague com os.remove()
                if os.path.isfile(item_path) or os.path.islink(item_path):
                    print(f"Apagando arquivo: {item_path}")
                    os.remove(item_path)
                # Se for um diretório, apague recursivamente com shutil.rmtree()
                elif os.path.isdir(item_path):
                    print(f"Apagando pasta e todo o seu conteúdo: {item_path}")
                    shutil.rmtree(item_path)
            except Exception as e:
                print(f"Erro ao apagar {item_path}. Motivo: {e}")

        print("\nLimpeza concluída com sucesso!")
    else:
        print("O caminho especificado não existe. Nenhuma ação foi realizada.")
```
